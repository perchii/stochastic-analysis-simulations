### Датчик схемы Бернулли. Датчик биномиального распределения 

<span style="color:cyan;">Определение. </span> Случайная величина $X$, принимающая значение $1$ с вероятностью $p$ и значение $0$ с вероятностью $q=1-p$, называется случайной величиной с распределением Бернулли.


Для получения генератора схемы Бернулли с заданной вероятностью успеха $p$ возьмем генератор равномерно распределенных на отрезке $[0,1]$ случайных величин. Сгенерируем равномерно распределенную случайную величину $X\sim U[0,1]$, тогда $Y \sim Bern (p) =\mathbb{I}(X<p)$, где $\mathbb{I}$ - индикаторная функция:
\begin{equation}
Y=\mathbb{I}(X<p)=\left\{
\begin{aligned}
&1, X<p,\\
&0, X\geqslant p.
\end{aligned}
\right.
\end{equation}

<span style="color:cyan;">Определение. </span> Случайная величина $X$ имеет биномиальное распределение с параметрами $n$ и $p$, $X\sim Bin(n,p)$, если
\begin{equation}
\mathbb{P}(X=k)=C^k_n p^k (1-p)^{n-k},\quad k\in \mathbb{N}\cup\{0\}.
\end{equation}

Случайную величину $X$ обычно интерпретируют как число успехов в серии из $n$ одинаковых независимых 
испытаний Бернулли с вероятностью успеха $p$ в каждом из них. Поэтому:
\begin{equation}
X = \sum_{i=1}^n Y_i,
\end{equation}
где $X\sim Bin(n,p),\quad Y\sim Bern(p)$.

```python
import random
import numpy as np

def bernoulli_gen(p, size=None):
    if not 0 <= p <= 1:
        raise ValueError("p должна быть между 0 и 1")

    if size is None:
        return int(np.random.rand() < p)

    u = np.random.rand(size)        
    return (u < p).astype(int)          
    


def bin_res(n, p, k):
    res = np.zeros(k)
    for i in range(k):
        res[i] = int(np.sum(bernoulli_gen(n, p)))
    return res

```

### Датчик геометрического распределения

<span style="color:cyan;">Определение. </span> Случайная величина $X$ имеет геометрическое распределение с параметром $p$, $X\sim Geom(p)$, если
$$
\mathbb{P}(X=k)=(1-p)^kp=q^k p,\quad k\in\mathbb{N}\cup\{0\}.
$$

Так же, как и в случае биномиального распределения, проводится некоторое количество испытаний Бернулли с одинаковой вероятностью успеха, до первого успеха. В качестве случайной величины с геометрическим распределением берем количество неудач до первого успеха.


```python
def geom_gen_p(p):
    k = 0
    f = 0
    while f != 1:
        f = bernoulli_gen(p)
        k += 1
    return k

def geom_res_p(n, p):
    return [geom_gen_p(p) for i in range(n)]

geom_res_p(10, 0.001)
```

Сделаем более эффектиыную версию генератора


Если $G \sim \mathrm{Geom}(p)$, то

$$
G = \left\lfloor \frac{\ln(1 - U)}{\ln(1 - p)} \right\rfloor + 1,
\quad U \sim \mathrm{Unif}(0, 1)
$$



```python
def geom_res(n, p):
    if not 0 < p <= 1:
        raise ValueError("p должна быть в (0,1]")
    
    u = np.random.rand(n)  
    g = np.floor(np.log(1 - u) / np.log(1 - p)) + 1
    return g.astype(int)  
```

### Свойство отсутствия памяти

<span style="color:cyan;">Определение. </span> Пусть $Y\sim Geom(p)$, тогда $\forall m,n\in\mathbb{N}\cup\{0\}$ справедливо:
$$
\mathbb{P}(\,Y\geqslant m+n\mid Y\geqslant m\,)=\mathbb{P}(Y\geqslant n),
$$
то есть количество прошлых неудач не влияет на количество будущих неудач.


Рассмотрим левую часть равенства:

$$\mathbb{P}(\,Y\geqslant m+n\mid Y\geqslant m\,)=\dfrac{\mathbb{P}(\,Y\geqslant m+n, Y\geqslant m\,)}{\mathbb{P}(Y\geqslant m)}=\\
=\dfrac{\mathbb{P}(Y\geqslant m+n)}{\mathbb{P}(Y\geqslant m)}=\dfrac{\sum_{i=m+n}^\infty q^i p}{\sum_{i=m}^\infty q^i p}=\dfrac{q^{m+n}}{q^m}=q^n.$$


С другой стороны, правая часть равна:
$$
\mathbb{P}(Y\geqslant n)=\sum_{i=n}^\infty q^i p=p\dfrac{q^n}{1-q}=q^n.
$$\\\\


### Игра в орлянку

Рассмотрим процесс игры в орлянку. Для этого смоделируем последовательность случайных величин $X_1, X_2, . . .,$ где
\begin{equation}
	X_n=\left\{
	\begin{aligned}
	&1, & \xi \in [0, 0.5)\\
	&-1, &\xi \in [0.5, 1]
	\end{aligned}
	\right.
\end{equation}
$\xi \sim U(0,1)$.
Построим траекторию процесса $Y(i)$
$$Y(i) = \frac{X_1 + \dots + X_i}{\sqrt{n}}$$

где $i = 1,n$



```python
import matplotlib.pyplot as plt
import numpy as np

def coin(p):
    if p > 1 or p < 0:
        raise ValueError("p должна быть между 0 и 1")
    if random.random() < p:
        return 1
    else:
        return -1  


K = 1000
S = np.zeros(K)
S[1] = coin(0.5)
for i in range(2, K):
    S[i] = S[i-1] + coin(0.5)
n = np.arange(1, K+1)
S = np.array(S)/np.sqrt(K)
plt.plot(n, S)
```

**Центральная предельная теорема** *(доказательство см. в [2])*

Пусть $X_1, \ldots, X_n, \ldots$ — бесконечная последовательность независимых одинаково распределённых случайных величин с конечными:
- Математическим ожиданием $\mu$
- Дисперсией $\sigma^2 > 0$

Обозначим частичную сумму:
$$
S_n = \sum_{i=1}^n X_i
$$

Тогда выполняется сходимость:
$$
\frac{S_n - \mu n}{\sigma \sqrt{n}} \xrightarrow{d} N(0,1) \quad \text{при} \quad n \to \infty
$$
где $N(0,1)$ — стандартное нормальное распределение (с нулевым средним и единичной дисперсией).

У нас математическое ожидание $X_i$ равно 0, а дисперсия равна 1. Следовательно:

$$\frac{Y(n)}{\sqrt{n}} \to N(0, 1)$$

### Датчик сингулярного распределения (Канторова лестница)

<span style="color:cyan;">Определение. </span> Функция распределения называется сингулярной, если она непрерывна и ее множество точек роста имеет нулевую меру Лебега.

Рассмотрим алгоритм построения канторова множества.
Из единичного отрезка $C_0=[0,1]$ удалим интервал $(1/3,2/3)$. Оставшееся точечное множество обозначим через $ C_1 $. Множество $ C_1=[0,1/3]\cup[2/3,1]$ состоит из двух отрезков; удалим теперь из каждого отрезка его среднюю треть, и оставшееся множество обозначим через $ C_2 $. Повторив эту процедуру опять, удаляя средние трети у всех четырёх отрезков, получаем $ C_3 $. Дальше таким же образом получаем последовательность замкнутых множеств $C_0\supset C_1\supset C_2\supset\ldots $ Пересечение
$$
C=\bigcap\limits_{i=0}^{+\infty}C_i
$$
называется канторовым множеством.


Канторово множество $K$ можно определить как множество чисел от нуля до единицы, представимое в троичной системе счисления с помощью нулей и двоек. 

Посчитаем меру Лебега канторова множества. Вычтем длины отрезков, которые мы выкинули, из длины целого отрезка $ [0,1] $:
$$ \mu(K)=1-\sum\limits_{k=1}^{+\infty}\dfrac{2^{k-1}}{3^k}=1-\dfrac{1}{3}\cdot\dfrac{1}{1-\frac{2}{3}}=0. $$
Таким образом, любая функция распределения, точки роста которой совпадает с точками канторова множества, является сингулярной. 
Как упоминалось ранее, точки канторова множества в троичной системе задаются только нулями и двойками, следовательно, все числа можно получить при помощи генератора случайной величины Бернулли следующим образом:
$$
x=\sum_{i=1}^\infty \dfrac{2\alpha_i}{3^i},\quad x\in K, 
$$
где $\alpha_i\sim Bern(0.5)$.
Суммирование в  ведется до бесконечности. При построении будем суммировать до некоторого разряда $n\in\mathbb{N}$.


```python
def singular_gen(n, k = 25):
    return np.array([np.sum(np.array([bernoulli_gen(0.5) * 2 / 3 ** i for i in range(1, k+1)])) for j in range(n)])
singular_gen(10)

```

### Проверка критерия Колмогорова


Для проверки корректности работы датчика предлагается воспользоваться критерием Колмогорова. Будем считать максимальное по модулю отклонение эмпирической функции распределения от теоретической в точках выборки. Обозначим это число через $ D_n $. Для проверки простой гипотезы $ H_0 $ о соответствии выборки известному закону распределения с заданным уровнем значимости $ \alpha $ рассчитаем $p-$ value по статистике $ \sqrt{n}D_n$, то есть найдём значение функции распределения Колмогорова в заданной точке. Для этого используем следующее представление для функции распределения Колмогорова:
$$
F_K(x)=1+2\sum\limits_{k=1}^{+\infty}(-1)^ke^{-2k^2x^2}
$$

Гипотеза принимается, если $ p=1-F_K(\sqrt{n}D_n) $ превышает заданный уровень значимости $ \alpha $.

```python
import matplotlib.pyplot as plt
import numpy as np

def cantor(n=1, eps=1e-10):
    m = int(np.round(-np.log(eps) / np.log(9)))
    bern = np.array([bernoulli_gen(m, 0.5) for i in range(n)])
    deg = -np.arange(1, m + 1)
    x = 2 * bern.dot(3.0 ** deg)
    F = bern.dot(2.0 ** deg)
    x_sort = np.sort(x)
    F_sort = np.sort(F)
    return x_sort, F_sort

def Empir_Cant(n, x):
    x_s = singular_gen(n, 100)
    return [np.mean([(x_s[i] < xx) for i in range(n)]) for xx in x]

def Empir_Cant_Sim(n, x):
    x_s = 1 - singular_gen(n, 100)
    return [np.mean([(x_s[i] < xx) for i in range(n)]) for xx in x]

def F_k(x, n):
    return 1 + 2 * np.sum([(-1) ** i * np.exp(-2 * i ** 2 * x ** 2) for i in range(1, n+1)])


alpha = 0.05 #уровень значимости
N = 1000 #кол-во запусков
procent = 0
for _ in range(N):
    n = 1000 #объем выборки
    x, F = cantor(n, 0.000000001)
    F_n = Empir_Cant(n, x)
    D_n = np.max(np.abs(F - F_n))
    p = -2 * np.sum([(-1) ** i * np.exp(-2 * i ** 2 * (np.sqrt(n) * D_n) ** 2) for i in range(1, 10000)])
    procent += (p > alpha)
print('Процент принятия гипотезы:', procent * 100 / N)
```

### Свойство симметричности и самоподобия

Покажем свойство симметричности. Пусть имеется канторова случайная величина $X~=~\sum_{i=1}^\infty \dfrac{2\alpha_i}{3^i}$, где $\alpha_i\sim Bern(0.5)$. Рассмотрим случайную величину $1-X$:
$$ 1-X=1-\sum_{i=1}^\infty \dfrac{2\alpha_i}{3^i}=\sum_{i=1}^\infty \dfrac{2}{3^i}-\sum_{i=1}^\infty \dfrac{2\alpha_i}{3^i}=\sum_{i=1}^\infty \dfrac{2(1-\alpha_i)}{3^i}=\sum_{i=1}^\infty \dfrac{2\beta_i}{3^i}. $$
Здесь $\beta_i\sim Bern(0.5)$, поэтому случайные величины $1-X$ и $X$ распределены одинаково.\\

Покажем свойство самоподобия относительно деления на $3$. Рассмотрим условное распредление канторовой случайной величины $Y$ на отрезке $\left[0;\dfrac{1}{3}\right]$. Это будет соответствовать тому, что $\alpha_1=0$. В таком случае:
$$ Y=\sum_{i=2}^\infty \dfrac{2\alpha_i}{3^i}=\sum_{i=1}^\infty \dfrac{2\alpha_{i+1}}{3^{i+1}}=\dfrac{1}{3}\sum_{i=1}^\infty \dfrac{2\alpha_i}{3^i}=\dfrac{1}{3}Y. $$

### Проверка критерия Смирнова

Проверим последние упомянутые свойства с помощью критерия Смирнова. Найдем максимум поточечной разности двух эмперических функций распределения. Обозначив полученное значение через $ D_{nm} $, где $ n,m $ - длины выборок, и посчитав $ p- $ value по статистике $\sqrt{\frac{nm}{n+m}}D_{nm}$ для заданного уровня значимости $ \alpha $ проверим гипотезу $ H_0 $ о том, что обе выборки соответствую одинаковому закону распределения.

(теорема Смирнова)
Пусть $ F_{n}^1(x) $, $ F_{m}^2(x) $ - эмпирические функции распределения с объёмами выборок $ n $ и $ m $ соответственно случайной величины $ \xi $. Тогда, если $ F(x)\in C^1(\mathbb{R}) $, то

$$\lim\limits_{n,m\to+\infty}\mathbb{P}\left(\sqrt{\dfrac{nm}{n+m}}D_{n,m}\leqslant t\right)=K(t)=\sum\limits_{-\infty}^{+\infty}(-1)^je^{-2j^2t^2}, \ \  	\forall t>0$$

где $ D_{n,m}=\sup\limits_{x\in\mathbb{R}}|F^1_{n}-F^2_{m}| $.

```python
def Empir_Cant(n, x):
    x_s = singular_gen(n, 100)
    return np.array([np.mean([(x_s[i] < xx) for i in range(n)]) for xx in x])

def Empir_Cant_Sim(n, x):
    x_s = 1 - singular_gen(n, 100)
    return np.array([np.mean([(x_s[i] < xx) for i in range(n)]) for xx in x])

alpha = 0.05
N = 1000
procent = 0
for t in range(N):
    n = 1000
    m = 2000
    x, F = cantor(n, 0.000000001)
    F_n1 = Empir_Cant(n, x)
    F_n2 = Empir_Cant_Sim(m, x)
    D_nm = np.max(np.abs(F_n2 - F_n1))
    p = -2 * np.sum([(-1) ** i * np.exp(-2 * i ** 2 * (np.sqrt(n * m / (n + m)) * D_nm) ** 2) for i in range(1, 100)])
    procent += (p > alpha)
print('Процент принятия гипотезы:', procent * 100 / N)
```

```python
def ber(p, n, m):
    return np.random.rand(n, m) < p

def empirical_cdf(sample):
    sorted_sample = np.sort(sample)
    n = len(sorted_sample)
    F = np.arange(1, n + 1) / n
    return sorted_sample, F

def cantor_2(n=1, eps=1e-10):
    m = int(round(-np.log(eps) / np.log(9)))
    bern = ber(0.5, n, m).astype(float)
    deg = -(np.arange(1, m + 1))
    powers_of_three = 3.0 ** deg
    x = 2 * bern.dot(powers_of_three)
    x1 = x / 3
    t_1, F_1 = empirical_cdf(x1)
    x2 = x[x < 1/3]
    t_2, F_2 = empirical_cdf(x2)

    return t_1, F_1, t_2, F_2

def smirnov(X_m, X_n):
    m = len(X_m)
    n = len(X_n)
    X_mn = np.concatenate((X_m, X_n))
    indices = np.argsort(X_mn)
    X_mn_sorted = X_mn[indices]
    delta_plus = np.concatenate((np.zeros(m), np.ones(n)))
    delta_plus_sorted = delta_plus[indices]

    i = np.arange(1, m + n + 1)

    S_plus = (i * n) / (m + n) - np.cumsum(delta_plus_sorted)
    D_mn_plus = (m + n) * np.max(np.abs(S_plus)) / (m * n)
    delta_minus = np.concatenate((np.ones(m), np.zeros(n)))
    delta_minus_sorted = delta_minus[indices]
    S_minus = (i * m) / (m + n) - np.cumsum(delta_minus_sorted)
    D_mn_minus = (m + n) * np.max(np.abs(S_minus)) / (m * n)
    return max(D_mn_plus, D_mn_minus)

alpha = 0.05
N = 1000
procent = 0
for t in range(N):
    x1, F1, x2, F2 = cantor_2(1000, 0.000000001)
    n = len(x1)
    m = len(x2)
    D_nm = smirnov(x1, x2)
    p = -2 * np.sum([(-1) ** i * np.exp(-2 * i ** 2 * (np.sqrt(n * m / (n + m)) * D_nm) ** 2) for i in range(1, 100)])
    procent += (p > alpha)
print('Процент принятия гипотезы:', procent * 100 / N)
```

### Сходимость эмпирических матожидания и дисперсии

Вычислим математическое ожидание и дисперсию рассматриваемой случайной величины. 
Как упоминалось ранее, $ F $ обладает свойством самоподобия, то есть при $  0<x<1/3 $ выполнено соотношение $ F(x)=F(3x)/2 $, а при $ 2/3<x<1 $ имеет место равенство $ F(x)=1/2+F(3x-2)/2 $. Поэтому
$$
\mathbb{E}\xi=\int\limits_{-\infty}^{+\infty}xdF(x)=\int\limits_{0}^{1/3}xdF(x)+\int\limits_{2/3}^{1}xdF(x)=\dfrac{1}{2}\int\limits_{0}^{1/3}xdF(3x)+\dfrac{1}{2}\int\limits_{2/3}^{1}xd(1/2+F(3x-2)).
$$
Далее введем замену $y=3x$ в первом интеграле и $y=3x-2$ во втором интеграле:

$$
\mathbb{E}\xi=\dfrac{1}{2}\int\limits_{0}^{1}\dfrac{y}{3}dF(y)+\dfrac{1}{2}\int\limits_{0}^{1}\dfrac{y+2}{3}dF(y)=\dfrac{1}{6}\int\limits_{0}^{1}ydF(y)+\dfrac{1}{6}\int\limits_{0}^{1}ydF(y)+\dfrac{1}{3}\int\limits_{0}^{1}dF(y)=\dfrac{1}{3}\mathbb{E}\xi+\dfrac{1}{3}.
$$
Таким образом, получаем  $ \mathbb{E}\xi=1/2 $.

Аналогичным способом с использованием свойства самоподобия вычислим дисперсию величины $ \xi $, используя вычисленное значения математического ожидания.
$$
\mathbb{E}\xi^2=\int\limits_{0}^{1/3}x^2dF(x)+\int\limits_{2/3}^{1}x^2dF(x)=\dfrac{1}{2}\int\limits_{0}^{1}\left(\dfrac{y}{3}\right)^2dF(y)+\dfrac{1}{2}\int\limits_{0}^{1}\left(\dfrac{y+2}{3}\right)^2dF(y)=
$$
$$
=\dfrac{1}{9}\mathbb{E}\xi^2+\dfrac{2}{9}\mathbb{E}\xi+\dfrac{2}{9}=\dfrac{1}{9}\mathbb{E}\xi^2+\dfrac{1}{9}+\dfrac{2}{9}.
$$
То есть имеем $ \mathbb{E}\xi^2=3/8 $. Таким образом, получаем значение дисперсии $ \mathbb{D}\xi=3/8-(1/2)^2=1/8 $.

### Датчик экспоненциального распределения 

Случайная величина $X$ имеет экспоненциальное распределение с параметром $\lambda > 0$, если ее функция распределения имеет вид:
$$
F_X(x) = \begin{cases}
1 - e^{-\lambda x}, &x \geqslant 0,\\
0, &x < 0.
\end{cases}
$$



Пусть на $\mathbb{R}$ определена функция $F(x)$ такая, что:
1. $F(x)$ непрерывна,
2. $F(x)$ монотонно возрастает,
3. $F(x) \to 0$ при $x \to -\infty$,
4. $F(x) \to 1$ при $x \to \infty$.

Пусть также задана случайная величина $Y\sim U[0, 1]$. Тогда функция $F(x)$ является функцией распределения случайной величины $X = F^{-1}(Y)$.

Функция распределения удовлетворяет условиям теоремы 1. Применим её для моделирования датчика экспоненциального распределения на основе датчика равномерного распределения:
$$
F_X(x)=1-e^{-\lambda x},
$$

$$
F_X^{-1}(y)=-\dfrac{1}{\lambda}\ln(1-y)
$$
и если $Y\sim U[0,1]$, то получаем случайную величину, имеющую экспоненциальное распределение с параметром $\lambda$:
$$
X=-\dfrac{1}{\lambda}\ln(1-Y)\sim Exp(\lambda).
$$

### Свойство отсутсвия памяти

Случайная величина $X\sim Exp(\lambda)$ обладает свойством отсутствия памяти, то есть $\forall t\ne 0$ и $\forall s$ следует, что

$$
\mathbb{P}(X\ge s+t\mid X\ge t)=\mathbb{P}(X\ge s). \tag{1}
$$

**Доказательство:**

Применим аксиому условной вероятности:

$$
\mathbb{P}(X\ge s+t\mid X\ge t)=\dfrac{\mathbb{P}(X\ge s+t,X\ge t)}{\mathbb{P}(X\ge t)}=\dfrac{\mathbb{P}(X\ge s+t)}{\mathbb{P}(X\ge t)}=\mathbb{P}(X\ge s). 
$$

Таким образом, получаем:

$$
\mathbb{P}(X\ge s+t)=\mathbb{P}(X\ge t)\mathbb{P}(X\ge s). \tag{2}
$$

Для экспоненциально распределённой случайной величины верно, что:

$$
\mathbb{P}(X\ge t)=1-F_X(t)=e^{-\lambda t},
$$

$$
\mathbb{P}(X\ge s+t)=e^{-\lambda(s+t)}.
$$

Следовательно, для (2) выполняется:

$$
e^{-\lambda(s+t)}=e^{-\lambda s}e^{-\lambda t}.
$$

Таким образом, экспоненциальное распределение обладает свойством отсутствия памяти.

### Случайная величина $Y = \min(X_1, X_2, \dots, X_n)$

**Утверждение:**  
Пусть $X_1,X_2,\ldots,X_n$ — независимые экспоненциально распределённые случайные величины с параметрами $\lambda_1,\lambda_2,\ldots,\lambda_n$ соответственно. Тогда случайная величина  
$$ Y = \min(X_1,X_2,\ldots,X_n) \sim \text{Exp}\left(\sum_{i=1}^n \lambda_i \right). $$

**Доказательство:**  
$$
\begin{aligned}
F_Y(x) &= \mathbb{P}(Y \leqslant x) = 1 - \mathbb{P}(Y > x) = 1 - \mathbb{P}(\min(X_1,X_2,\ldots,X_n) > x) \\
&= 1 - \mathbb{P}(X_1 > x, X_2 > x, \ldots, X_n > x) \\
&= 1 - \mathbb{P}(X_1 > x) \cdot \mathbb{P}(X_2 > x) \cdot \ldots \cdot \mathbb{P}(X_n > x) \quad (\text{т.к. } X_i \text{ независимы}) \\
&= 1 - \left(1 - F_{X_1}(x)\right) \left(1 - F_{X_2}(x)\right) \ldots \left(1 - F_{X_n}(x)\right) \\
&= 1 - e^{-\lambda_1 x} \cdot e^{-\lambda_2 x} \cdot \ldots \cdot e^{-\lambda_n x} \\
&= 1 - e^{-\left(\sum_{i=1}^n \lambda_i \right) x}.
\end{aligned}
$$  
Таким образом, $Y$ имеет экспоненциальное распределение с параметром $\sum_{i=1}^n \lambda_i$.

### Датчик пуассоновского распределения

**Определение:**  
Случайная величина $X$ имеет распределение Пуассона с параметром $\lambda>0$, если  
$$
\mathbb{P}(X=k)=\dfrac{\lambda^k}{k!}e^{-\lambda},\quad k\in\mathbb{N}\cup\{0\}.
$$

**Теорема:**  
Пусть $X_1,X_2,\ldots,X_n,\ldots\sim Exp(\lambda)$ — независимые одинаково распределенные случайные величины. Тогда случайная величина, определенная следующим образом:  
$$
Y=\max_{n\in\mathbb{N}}\,(S_n=X_1+X_2+\ldots+X_n)<1
$$  
имеет распределение Пуассона с параметром $\lambda$ (при этом $Y=0$, если $X_1\ge 1$).

Таким образом, пуассоновскую случайную величину $Y=n-1$, $Y\sim Pois(\lambda)$ можно получить, генерируя экспоненциально распределенные случайные величины $X_i\sim Exp(\lambda)$ до тех пор, пока их сумма $S_n$ не станет превышать $1$.

```python
def pois_gen(l, n):
    pois = np.zeros(n)
    for i in range(n):
        sum = 0
        while sum < 1:
            pois[i] = sum
            sum += exp_gen(l, 1)
    return pois
pois_gen(2, 10)
```

### Проверка критерия хи-квадрат

**Определение (Сходимость по вероятности):**  
Последовательность случайных величин $\{\xi_n\}_{n=1}^{+\infty}$ сходится по вероятности (по мере $\mathbb{P}$) к величине $\xi$, если:
$$
\forall \varepsilon>0:\ \mathbb{P}\left(\left|\xi_n-\xi\right|\ge\varepsilon\right)\xrightarrow[n\to\infty]{} 0.
$$
Обозначение: $\xi_n\xrightarrow[n\to\infty]{\mathbb{P}}\xi$.

**Моделирование Пуассоновского распределения:**  
Другой способ моделирования пуассоновской случайной величины основан на предельном свойстве биномиального распределения:
$$
Bi(n,p)\xrightarrow[n\to\infty]{\mathbb{P}} Pois(\lambda),\quad \text{где } p=\dfrac{\lambda}{n}\xrightarrow[n\to\infty]{} 0.
$$

**Определение (Распределение хи-квадрат):**  
Пусть случайные величины $\xi_1,\xi_2,\ldots,\xi_n$ независимы и имеют стандартное нормальное распределение $\mathcal{N}(0,1)$. Тогда случайная величина:
$$
\chi_n^2=\xi_1^2+\xi_2^2+\ldots+\xi_n^2
$$
имеет хи-квадрат распределение с $n$ степенями свободы.

**Критерий хи-квадрат Пирсона:**  
Для дискретной неотрицательной случайной величины $X$ с распределением:
$$
\mathbb{P}(X=k)=p_k,\quad k\in\mathbb{N}\cup\{0\}
$$
статистика критерия вычисляется следующим образом:
1. $k_0$ - максимальное значение в выборке из $n$ испытаний
2. $n_k$ - количество элементов, принявших значение $k$
3. Статистика критерия:
$$
X_n^2=n\sum_{k=1}^{k_0} \dfrac{\left(\frac{n_k}{n}-p_k\right)^2}{p_k}
$$

**Проверка распределения Пуассона:**  
1. Генерируем $n$ пуассоновских величин, получаем эмпирические частоты $n_k$
2. Вычисляем теоретические вероятности:
$$
p_k=\mathbb{P}(X=k)=\dfrac{\lambda^k}{k!}e^{-\lambda},\quad k=0,1,\ldots,k_0
$$
3. Вычисляем статистику:
$$
\chi_{k_0}^2=\sum_{k=0}^{k_0}\dfrac{\left(n_k-np_k\right)^2}{np_k}
$$
4. Сравниваем с критическим значением $\chi^2_{\alpha,k_0}$
5. Если $\chi_{k_0}^2 < \chi^2_{\alpha,k_0}$, гипотеза $H_0$ о пуассоновском распределении не отвергается

```python
from scipy.stats import chi2

N = 100
pr = 0
for j in range(N):
    n_0 = 1000 #выборка длины n_0 с биномиальным арспределением
    n_b = 1000 #для экспериментов в биномальном распределение 
    n_k = np.zeros(n_b)
    lm = 1

    b = bin_res(n_b, lm/n_b, n_0)
    k0 = int(np.max(b))
    n_k = np.zeros(k0+1)
    for i in range(n_0):
        n_k[int(b[i])] += 1

    p = np.array([ lm ** k / np.math.factorial(k) * np.exp(-lm) for k in range(k0+1)])

    xi_2 = n_0 * np.sum([ (n_k[k]/n_0 - p[k]) ** 2 / p[k] for k in range(k0+1) ])
    alpha = 0.05
    critical_value = chi2.ppf(1 - alpha/2, k0-1)
    pr += (xi_2 <= critical_value)

print(pr * 100 / N)
```

### Датчик стандартного нормального распределения

**Определение (Нормальное распределение):**  
Случайная величина $\xi$ имеет нормальное распределение с параметрами $\mu$ и $\sigma^2$ ($\xi\sim\mathcal{N}(\mu,\sigma^2)$), если ее плотность:
$$
p_{\xi}(x)=\dfrac{1}{\sqrt{2\pi}\sigma}\exp\left(-\dfrac{(x-\mu)^2}{2\sigma^2}\right),\quad -\infty<x<+\infty.
$$

**Определение (Стандартное нормальное распределение):**  
При $\mu=0$ и $\sigma^2=1$ плотность имеет вид:
$$
p_{\xi}(x)=\dfrac{1}{\sqrt{2\pi}}\exp\left(-\dfrac{x^2}{2}\right),\quad -\infty<x<\infty.
$$

**Моделирование стандартного нормального распределения:**  
Для двумерного случая $X=(X_1,X_2)\sim \mathcal{N}(0,1)$ совместное распределение:
$$
\mathbb{P}(X_1<x_1,X_2<x_2)=\int\limits_{-\infty}^{x_1} \int\limits_{-\infty}^{x_2} \dfrac{1}{2\pi}e^{-\dfrac{\xi^2+\eta^2}{2}}d\xi d\eta.
$$

**Переход к полярным координатам:**  
$$
\begin{cases}
\xi=\rho\cos\phi \\
\eta=\rho\sin\phi
\end{cases}, \quad
J=\det\begin{pmatrix}
\cos\phi & -\rho\sin\phi \\
\sin\phi & \rho\cos\phi
\end{pmatrix}=\rho
$$

**Преобразование интеграла:**  
$$
\mathbb{P}(X_1<x_1,X_2<x_2)=\int\limits_{\substack{\rho\cos\phi<x_1\\ \rho\sin\phi<x_2}} \dfrac{1}{2\pi}\rho e^{-\dfrac{\rho^2}{2}} d\rho d\phi
$$

**Замена переменных ($\omega=\rho^2$):**  
$$
=\int\limits_{\substack{\sqrt{\omega}\cos\phi<x_1\\ \sqrt{\omega}\sin\phi<x_2}} \dfrac{1}{2}e^{-\dfrac{\omega}{2}}\dfrac{1}{2\pi} d\omega d\phi
$$

**Интерпретация:**  
Подынтегральное выражение - произведение плотностей:
- $Y_1\sim Exp\left(\frac{1}{2}\right)$
- $Y_2\sim U[0,2\pi]$

Таким образом:
$$
(X_1,X_2) \stackrel{d}{=} (\sqrt{Y_1}\cos Y_2, \sqrt{Y_1}\sin Y_2)
$$

**Независимость компонент:**  
$$
\mathbb{P}(X_1<x_1,X_2<x_2)=\left(\int\limits_{-\infty}^{x_1} \dfrac{1}{\sqrt{2\pi}} e^{-\dfrac{\xi^2}{2}} d\xi\right) \left(\int\limits_{-\infty}^{x_2} \dfrac{1}{\sqrt{2\pi}} e^{-\dfrac{\eta^2}{2}} d\eta\right)
$$

```python
def standart_norm_gen_sin(n):
    u = np.random.uniform(0, 1, n)
    return np.sqrt(exp_gen(1/2, n)).T * np.sin(2*np.pi*u)

def standart_norm_gen_cos(n):
    u = np.random.uniform(0, 1, n)
    return np.sqrt(exp_gen(1/2, n)).T * np.cos(2*np.pi*u)

standart_norm_gen_sin(10)
```

### Проверка t-критерия Стьюдента

**Проверка равенства математических ожиданий с использованием t-критерия Стьюдента**

Пусть:
- $M_1$ и $M_2$ - математические ожидания двух выборок
- $\overline{\xi}$ и $\overline{\eta}$ - выборочные средние
- $\Delta = \overline{\xi} - \overline{\eta}$ - разность выборочных средних

**Нулевая гипотеза ($H_0$):**  
$M_1 = M_2$, следовательно $\mathbb{E}(\Delta) = 0$

**Дисперсии:**  
При условии $\mathbb{D}(\xi) = \mathbb{D}(\eta) = 1$ и независимости наблюдений:
$$
\mathbb{D}(\overline{\xi}) = \frac{\mathbb{D}\left(\sum_{i=1}^n \xi_i\right)}{n^2} = \frac{1}{n} = \mathbb{D}(\overline{\eta})
$$
$$
\mathbb{D}(\Delta) = \frac{2}{n}
$$

**Оценки дисперсий:**  
Несмещённые оценки:
$$
s_1^2 = \frac{\sum_{i=1}^n (\xi_i - \overline{\xi})^2}{n-1}, \quad s_2^2 = \frac{\sum_{i=1}^n (\eta_i - \overline{\eta})^2}{n-1}
$$
Оценка дисперсии разности:
$$
s_{\Delta}^2 = \frac{s_1^2 + s_2^2}{n}
$$

**t-статистика:**  
$$
t = \frac{\overline{\xi} - \overline{\eta}}{\sqrt{s_1^2 + s_2^2}} \sqrt{n}
$$

**Правило принятия решения:**  
Если $|t| > t_{\alpha/2, n-1}$ (критическое значение распределения Стьюдента с $n-1$ степенями свободы), то нулевая гипотеза отвергается на уровне значимости $\alpha$.

```python
from scipy.stats import t

N = 1000 #запуски
n_size = 1000 # рамер выборок
alpha = 0.05
procent = 0

for i in range(N):
    ksi = standart_norm_gen_sin(n_size)
    eta = standart_norm_gen_cos(n_size)
    Eksi = np.mean(ksi)
    Eeta = np.mean(eta)



    s1 = np.sum([(ksi[i] - Eksi) ** 2 for i in range(n_size)]) / (n_size - 1)
    s2 = np.sum([(eta[i] - Eeta) ** 2 for i in range(n_size)]) / (n_size - 1)
    tt = (Eksi - Eeta) / np.sqrt(s1 + s2) * np.sqrt(n_size)

    t_crit = t.ppf(1 - alpha/2, n_size - 1)
    procent += (tt < t_crit)

print(procent * 100 / N)
```

### Проверка критерия Фишера

Для ранее вычисленных несмещённых оценок дисперсий:
$$
s_1^2 = \frac{\sum_{i=1}^n (\xi_i - \overline{\xi})^2}{n-1}, \quad s_2^2 = \frac{\sum_{i=1}^n (\eta_i - \overline{\eta})^2}{n-1}
$$

Статистика Фишера имеет вид:
$$
F = \frac{s_1^2}{s_2^2}
$$

**Гипотезы:**
- Нулевая гипотеза $H_0$: $\mathbb{D}(\xi) = \mathbb{D}(\eta)$
- Альтернативная гипотеза $H_1$: $\mathbb{D}(\xi) \neq \mathbb{D}(\eta)$

**Правило принятия решения:**
Нулевая гипотеза отвергается, если:
$$
F < F_{\alpha/2}(n-1,n-1) \quad \text{или} \quad F > F_{1-\alpha/2}(n-1,n-1)
$$
где:
- $F_{\alpha}(n-1,n-1)$ - $\alpha$-квантиль распределения Фишера
- $n$ - объём выборки
- $\alpha$ - уровень значимости

**Интерпретация:**
Если вычисленное значение $F$ попадает в критическую область, мы отвергаем гипотезу о равенстве дисперсий с уровнем значимости $\alpha$.

```python
from scipy.stats import f

N = 1000 #запуски
n_size = 1000 # рамер выборок
alpha = 0.95
procent = 0

for i in range(N):
    ksi = standart_norm_gen_sin(n_size)
    eta = standart_norm_gen_cos(n_size)
    Eksi = np.mean(ksi)
    Eeta = np.mean(eta)

    s1 = np.sum([(ksi[i] - Eksi) ** 2 for i in range(n_size)]) / (n_size - 1)
    s2 = np.sum([(eta[i] - Eeta) ** 2 for i in range(n_size)]) / (n_size - 1)

    ff = s1/s2


    f_low = f.ppf(alpha / 2, n_size - 1, n_size - 1)
    f_high = f.ppf(1 - alpha / 2, n_size - 1, n_size - 1)

    # гипотеза НЕ отвергается, если F-попадает в допустимый интервал
    if f_low < ff < f_high:
        pass 
    else:
        procent += 1  

print(procent * 100 / N)
```

```python
def f(t):
    tg = np.tan(np.pi / 2 * t)
    tg2 = tg ** 2

    tg_sum = np.sum(tg2)
    tg_prod = np.prod(tg2)
    cos2  = np.cos(np.pi / 2 * t) ** 2
    cos_prod = np.prod(cos2)

    return np.exp(-(tg_sum + 1 / (2 ** 7 * tg_prod))) / (tg_prod * cos_prod)

def quadrature(N, d=10):
    t = np.array([i * 2 / N - 1 for i in range(0, N + 1)])
    y = np.array([(t[i] + t[i - 1]) / 2 for i in range(1, N + 1)])
    I = 0
    for idx in np.ndindex(*(N,) * d):
        t_new = y[list(idx)]
        I += f(t_new)
    return I * (np.pi / N) ** d
```

## Метод случайного поиска

Для поиска минимума функции $f$ будем $N$ раз разыгрывать случайные величины
$x_1$ и $x_2$, считать для каждой пары значение $f(x_1,x_2)$ и затем из
полученных $N$ значений выбирать минимальное. Величины $x_1$ и $x_2$ будем
получать через полярные координаты, см. их распределение на единичном круге $A$:
$$
  \frac{1}{\pi}\iint\limits_{x_1^2+x_2^2\le 1} dx_1\,dx_2
  = \frac{1}{\pi}\int_0^1 r\,dr \int_0^{2\pi} d\varphi
  = \int_0^1 dr^2 \int_0^{2\pi} \frac{1}{2\pi}\,d\varphi .
$$

Видно, что их совместное распределение совпадает с совместным распределением
случайных величин $\varphi\sim U[0;2\pi]$ и $R$, имеющей функцию распределения

  $$F_R(x) =
  \begin{cases}
    0,   & x < 0,\\
    x^2, & x \in [0;1],\\
    1,   & x > 1.
  \end{cases}$$


Моделирование $R$ можно произвести с помощью метода обратных функций:

  $$F_R^{-1}(x) = \sqrt{x}, \qquad x \in [0;1],$$

откуда при $Y\sim U[0;1]$ имеем
  $$X = \sqrt{Y} \sim R.$$

$$
\begin{cases}
x_1 = R \cos \varphi, \\
x_2 = R \sin \varphi,
\end{cases}
$$
$$
\quad R \sim U[0,1],  \varphi \sim U[0,2\pi].
$$



```python
import math
import random

def f_function(x1, x2):
    eps = 1e-8

    if abs(x1) > eps:
        term1 = x1 ** 3 * math.sin(1 / x1)
    else:
        term1 = 0   # предел при x1 -> 0 равен 0
    if abs(x2) > eps:
        term2 = 10 * x1 * (x2 ** 4) * math.cos(1 / x2)
    else:
        term2 = 0   # предел при x2 -> 0 тоже 0

    return term1 + term2


def random_point_in_unit_disk():
    u = random.random()    
    r = math.sqrt(u)
    phi = 2 * math.pi * random.random()

    x1 = r * math.cos(phi)
    x2 = r * math.sin(phi)

    return x1, x2


def random_search_min(n):
    best_f = None
    best_x1 = None
    best_x2 = None
    for i in range(n):
        x1, x2 = random_point_in_unit_disk()
        value = f_function(x1, x2)

        if best_f is None or value < best_f:
            best_f = value
            best_x1 = x1
            best_x2 = x2
    tol = 51 * np.sqrt(0.99/n)
    return best_x1, best_x2, best_f, tol
```

В первом случае (когда минимум внутри $A$):

$$
p = 1 - \left(1 - \frac{2\pi \delta^2}{\pi\cdot 1^2}\right)^N
    = 1 - (1 - 2\delta^2)^N .
$$

Во втором случае (когда минимум на границе):

$$
p = 1 - \left(1 - \frac{2\cdot \frac12 \pi \delta^2}{\pi\cdot 1^2}\right)^N
    = 1 - (1 - \delta^2)^N .
$$

Таким образом, в худшем случае:

$$
\delta = \sqrt{1 - (1 - p)^{1/N}}
\approx \sqrt{\frac{p}{N}} .
$$

И окончательное неравенство погрешности:

$$
\bigl| f(x^{*},y^{*}) - f(x,y) \bigr|
\le 51 \sqrt{\frac{p}{N}} .
$$


### Метод имитации отжига
Метод имитации отжига — стохастический алгоритм поиска глобального минимума.  
Он имитирует процесс охлаждения металла: при высокой температуре система свободно перемещается по пространству состояний, а при охлаждении стабилизируется в зоне минимума.


Состояние:
$$
s_i = (x_i, y_i)
$$

Температура:
$$
T_i
$$

Минимизируемая функция (функция Розенброка):
$$
g(x,y) = (x-1)^2 + 100 (y - x^2)^2
$$


Новое состояние получается добавлением нормального случайного шага.  

Шаг:
$$
x_{i+1} = x_i + \sigma \sqrt{T_i}\, Z_1
$$
$$
y_{i+1} = y_i + \sigma \sqrt{T_i}\, Z_2
$$

где  
$Z_1, Z_2 \sim N(0,1)$.


$$
\Delta g = g(x_{i+1}, y_{i+1}) - g(x_i, y_i)
$$


### Правило приёма нового состояния

Если $\Delta g < 0$, принимаем новое состояние.

Если $\Delta g > 0$, принимаем с вероятностью:
$$
P = \exp\!\left( - \frac{\Delta g}{T_i} \right)
$$


Температура уменьшается по закону:
$$
T_{i+1} = k\, T_i, \qquad 0 < k < 1
$$


### Условие остановки

Алгоритм завершается, когда
$$
T_i < \varepsilon
$$
или достигнуто максимальное число итераций.


### Алгоритм 

1. Задать параметры:  
   начальное состояние $s_0 = (x_0, y_0)$,  
   температуру $T_0$, коэффициент охлаждения $k$,  
   масштаб $\sigma$, точность $\varepsilon$.

2. Пока $T_i \ge \varepsilon$:

   - Сгенерировать кандидата:
     $$
     s_{i+1} = s_i + \sigma \sqrt{T_i}\,(Z_1, Z_2)
     $$
   - Вычислить:
     $$
     \Delta g = g(s_{i+1}) - g(s_i)
     $$
   - Если $\Delta g < 0$, принять новое состояние.
   - Иначе принять с вероятностью:
     $$
     P = \exp\!\left( -\frac{\Delta g}{T_i} \right)
     $$
   - Обновить температуру:
     $$
     T_{i+1} = k T_i
     $$

3. Вернуть лучшее найденное состояние


```python
def g_function(x, y):
    return (x - 1) ** 2 + 100 * (y - x ** 2) ** 2


def simulated_annealing(x0, y0, T0, k, sigma, eps):
    x = x0
    y = y0
    T = T0

    best_x = x
    best_y = y
    best_g = g_function(x, y)

    i = 0
    while T >= eps:
        std = sigma * math.sqrt(T)

        z = standart_norm_gen_cos(2)
        x_new = x + std * z[0]
        y_new = y + std * z[1]

        g_old = g_function(x, y)
        g_new = g_function(x_new, y_new)
        delta_g = g_new - g_old

        if delta_g < 0:
            accept = True
        else:
            P = math.exp(-delta_g / T)
            accept = (np.random.uniform(0, 1) < P)

        if accept:
            x = x_new
            y = y_new
            if g_new < best_g:
                best_g = g_new
                best_x = x
                best_y = y
        T = k * T
        i += 1

    return best_x, best_y, best_g, i



np.random.seed(0)

x0 = 0
y0 = 0
T0 = 2
k = 0.995
sigma = 2.7
eps = 1e-3

bx, by, bg, iters = simulated_annealing(x0, y0, T0, k, sigma, eps)

print("Итераций:", iters)
print("Лучшее значение g =", bg)
print("Точка (x, y) =", bx, by)
```

## Решение методом Монте–Карло

Строим квадратную сетку на области:
$$
[-1,1] \times [-1,1]
$$
с шагом h. Узел сетки (i,j) считается внутренним, если все его четыре соседа также лежат внутри круга:
$$
x_i^2 + y_j^2 \le 1.
$$
Если хотя бы один сосед вне круга, узел считается граничным.

Для граничных узлов значение решения известно из условия:
$$
u(i,j) = f(x_i, y_j) = x_i^2 - y_j^2.
$$


### Метод случайного блуждания

Для каждого внутреннего узла запускаем N траекторий случайного блуждания:

- из узла (i,j) выбираем равновероятно одного из соседей:  
  (i-1,j), (i+1,j), (i,j-1), (i,j+1);
- продолжаем блуждание, пока не попадаем на граничный узел;
- фиксируем значение f в точке выхода.

Пусть за s-ой траекторией выход произошёл в узле с номером $k_s$, где  
значение граничной функции равно $f_{k_s}$.  
Тогда оценка решения в узле (i,j) равна:
$$
u_{i,j} = \frac{1}{N} \sum_{s=1}^{N} f_{k_s}.
$$

Таким образом, решение в каждой внутренней точке — это среднее значение функции на границе, полученное по траекториям блужданий.

```python
sigma = 1
theta = 0.01
N = 2000
h = 1 / N

a = np.exp(-theta * h)
q = sigma**2 * (1 - np.exp(-2 * theta * h))

print("a =", a)
print("q =", q)

x = np.zeros(N)
x[0] = np.random.normal(0, sigma)
for n in range(N - 1):
    nu = np.random.normal(0, np.sqrt(q))
    x[n+1] = a * x[n] + nu
r = 0.05
eps = np.random.normal(0, np.sqrt(r), size=N)
y = x + eps

def kalman_filter(y, a, q, r, sigma_init):
    N = len(y)
    x_hat = np.zeros(N)
    P = np.zeros(N)
    x_hat[0] = 0
    P[0] = sigma_init**2

    for n in range(1, N):
        x_pred = a * x_hat[n-1]
        P_pred = a*a * P[n-1] + q
        K = P_pred / (P_pred + r)
        x_hat[n] = x_pred + K * (y[n] - x_pred)
        P[n] = (1 - K) * P_pred
    return x_hat, P

x_hat, P = kalman_filter(y, a, q, r, sigma_init=sigma)
M = 400
t = np.arange(N) * h

plt.figure(figsize=(12, 6))
plt.plot(t[:M], x[:M], label="Истинный процесс x_n", linewidth=2)
plt.scatter(t[:M], y[:M], s=8, alpha=0.5, label="Зашумлённые наблюдения y_n", marker='o')
plt.plot(t[:M], x_hat[:M], label="Оценка фильтра Калмана $\hat{x}_n$", linewidth=2)
plt.legend()
plt.title("Фильтрация процесса Орнштейна–Уленбека (белый шум)")
plt.xlabel("t")
plt.grid(True)
plt.show()
```

# Задание 11

Требуется построить двумерное пуассоновское поле, соответствующее работе одноканальной системы массового обслуживания (СМО). Каждое событие в такой системе характеризуется двумя величинами:

- временем поступления заявки;
- временем её обслуживания.

Таким образом, каждая заявка задаёт точку в двумерном пространстве  
$$
(t_j, s_j),
$$
где $t_j$ — момент прихода, а $s_j$ — длительность обслуживания.


Число заявок: 
$$
\eta \sim \mathrm{Pois}(\lambda T),
$$
где $\lambda > 0$ — интенсивность потока.

Условно при фиксированном числе заявок моменты их наступления распределены равномерно:
$$
t_1, \ldots, t_\eta \sim U(0, T),
$$
после чего их сортируют:
$$
t_{(1)} < t_{(2)} < \dots < t_{(\eta)}.
$$

Время обработки каждой заявки моделируется независимой случайной величиной
$$
s_j \sim \chi^2(10), \qquad j = 1, \ldots, \eta.
$$

Так как $\mathbb{E}[s_j] = 10$, средняя скорость обслуживания равна
$$
\mu = \frac{1}{10} = 0.1.
$$


Обозначим через $X_j$ момент, когда заявка полностью обработана. Тогда
$$
X_1 = t_{(1)} + s_1,
$$
а для $j \ge 2$
$$
X_j =
\begin{cases}
t_{(j)} + s_j, & \text{если } X_{j-1} \le t_{(j)},\\[4pt]
X_{j-1} + s_j, & \text{если } X_{j-1} > t_{(j)}.
\end{cases}
$$

Это означает, что если сервер свободен к моменту прихода заявки, она обслуживается сразу или иначе становится в очередь и ждёт окончания предыдущей.


Функция
$
N(t)
$
определяется как
$$
N(t) = \sum_{j=1}^{\eta} \mathbf{1}_{\{\, t_{(j)} \le t < X_j \,\}}.
$$

То есть заявка учитывается в системе, пока она уже поступила ($t_{(j)} \le t$) и ещё не завершила обслуживание ($t < X_j$).

Так как среднее время обслуживания равно $10$, критическая интенсивность равна
$$
\lambda_c = \frac{1}{10} = 0.1.
$$

- При $\lambda < 0.1$ система справляется с нагрузкой и очередь остаётся ограниченной.  
- При $\lambda = 0.1$ система работает на границе устойчивости.  
- При $\lambda > 0.1$ очередь имеет тенденцию к росту, так как заявок поступает больше, чем система успевает обслужить.


```python
import numpy as np
import matplotlib.pyplot as plt

def simulate_queue(lambda_, T, df_chi2, n):
    eta = np.random.poisson(lambda_ * T)

    if eta == 0:
        t_sorted = np.array([])
        s = np.array([])
        X = np.array([])
        t_grid = np.linspace(0, T, n)
        N_t = np.zeros_like(t_grid, dtype=int)
        return t_sorted, s, X, t_grid, N_t
    t = np.random.uniform(0, T, size=eta)
    t_sorted = np.sort(t)
    s = np.random.chisquare(df=df_chi2, size=eta)
    X = np.zeros(eta)
    X[0] = t_sorted[0] + s[0]
    for j in range(1, eta):
        if X[j - 1] <= t_sorted[j]:
            X[j] = t_sorted[j] + s[j]
        else:
            X[j] = X[j - 1] + s[j]
    t_grid = np.linspace(0.0, T, n)
    N_t = np.zeros_like(t_grid, dtype=int)

    for j in range(eta):
        mask = (t_grid >= t_sorted[j]) & (t_grid < X[j])
        N_t[mask] += 1

    return t_sorted, s, X, t_grid, N_t

T = 200
df_chi2 = 10
lambdas = [0.05, 0.1, 0.2]
plt.figure(figsize=(12, 8))
for i, lam in enumerate(lambdas, start=1):
    _, _, _, t_grid, N_t = simulate_queue(
        lambda_=lam,
        T=T,
        df_chi2=df_chi2,
        n=2000
    )
    plt.subplot(3, 1, i)
    plt.step(t_grid, N_t, where='post')
    plt.xlabel("t")
    plt.ylabel("N(t)")
    plt.title(f"Размер очереди N(t) при λ = {lam}")
    plt.grid(True, linestyle="--", alpha=0.5)

plt.tight_layout()
plt.show()
```

Рассмотрим неоднородный пуассоновский процесс с циклической интенсивностью вида
$$
\lambda(t) = \lambda_0 \bigl(1 + \cos t\bigr),
$$
где $\lambda_0 > 0$ — базовый уровень интенсивности. Интенсивность максимальна, когда $\cos t = 1$, и минимальна, когда $\cos t = -1$, то есть поток заявок усиливается и ослабевает периодически.
Для моделирования такого процесса используется метод Льюиса–Шедлеара.
Находим верхнюю оценку интенсивности:
$$
\lambda^* = \max_t \lambda(t) = 2 \lambda_0.
$$
Сначала моделируем однородный пуассоновский процесс с интенсивностью $\lambda^*$ на отрезке $[0, T]$:
$$
\eta^* \sim \mathrm{Pois}(\lambda^* T),
$$
а времена кандидатов равномерно распределены:
$$
t_1^*, \ldots, t_{\eta^*}^* \sim U(0, T).
$$
Для каждой кандидатной точки $t_i^*$ принимаем её как реальное событие с вероятностью
$$
p(t_i^*) = \frac{\lambda(t_i^*)}{\lambda^*}
= \frac{\lambda_0 (1 + \cos t_i^*)}{2 \lambda_0}
= \frac{1 + \cos t_i^*}{2}.
$$
То есть генерируется случайная величина $U_i \sim U(0,1)$, и точка принимается, если
$$
U_i < p(t_i^*).
$$
Оставшиеся точки образуют неоднородный пуассоновский процесс с интенсивностью $\lambda(t)$.
Каждому событию сопоставляется пара
$$
(t_j, x_j),
$$
где $t_j$ — время события, а $x_j$ — величина скачка. Вторая координата определяется как бернуллиевская случайная величина с параметром $p = 1$, то есть
$$
x_j \equiv 1.
$$

Таким образом, каждое событие увеличивает процесс на единицу, и двумерное поле имеет вид
$$
(t_j, 1), \quad j = 1, \ldots, N,
$$
где $N$ — число оставленных событий.

Число событий, произошедших к моменту времени $t$, задаётся как
$$
N(t) = \sum_{j=1}^{N} \mathbf{1}_{\{ t_j \le t \}}.
$$

Функция $N(t)$ является ступенчатой, возрастает только в моменты наступления событий и отражает неравномерность интенсивности: на интервалах, где $\lambda(t)$ велика, $N(t)$ растёт быстрее; там же, где $\lambda(t)$ мала, почти не изменяется.


```python
import numpy as np
import matplotlib.pyplot as plt


def simulate_nonhom_poisson(lambda0, T, n):
    lmb = 2 * lambda0
    eta = np.random.poisson(lmb * T)
    if eta == 0:
        t_events = np.array([])
        t_grid = np.linspace(0, T, n)
        N_t = np.zeros_like(t_grid, dtype=int)
        return t_events, t_grid, N_t
    t_candidates = np.random.uniform(0, T, size=eta)
    probs = (1 + np.cos(t_candidates)) / 2
    u = np.random.uniform(0, 1, size=eta)

    mask = u < probs
    t_events = t_candidates[mask]
    t_events = np.sort(t_events)
    t_grid = np.linspace(0, T, n)
    N_t = np.zeros_like(t_grid, dtype=int)
    idx = np.searchsorted(t_events, t_grid, side="right")
    N_t = idx
    return t_events, t_grid, N_t


T = 50
lambda0 = 5
np.random.seed(0)
t_events, t_grid, N_t = simulate_nonhom_poisson(lambda0, T, 2000)
lambda_t = lambda0 * (1 + np.cos(t_grid))
plt.figure(figsize=(12, 8))


plt.subplot(2, 1, 1)
plt.plot(t_grid, lambda_t)
plt.xlabel("t")
plt.ylabel("λ(t)")
plt.title("Интенсивность $\lambda(t) = \lambda_0 (1 + \cos t)$")
plt.grid(True, linestyle="--", alpha=0.5)


plt.subplot(2, 1, 2)
plt.step(t_grid, N_t, where="post")
plt.xlabel("t")
plt.ylabel("N(t)")
plt.title("Число событий N(t)")
plt.grid(True, linestyle="--", alpha=0.5)

plt.tight_layout()
plt.show()
```

Рассмотрим стохастическую модель капитала страховой компании. Поток страховых случаев задаётся пуассоновским процессом, а величина ущерба при каждом случае является случайной величиной с распределением Парето. Необходимо моделировать капитал компании во времени и анализировать, в каких условиях наступает разорение.

Считается, что страховые события происходят по однородному пуассоновскому процессу с интенсивностью $\lambda > 0$. Тогда число событий на промежутке $[0, T]$ имеет распределение
$$
\eta \sim \mathrm{Pois}(\lambda T),
$$
а времена наступления событий равномерны при условии известного числа событий:
$$
t_1, \ldots, t_\eta \sim U(0, T),
$$
после чего они сортируются:
$$
t_{(1)} < t_{(2)} < \dots < t_{(\eta)}.
$$



Величина ущерба при $j$-м страховом случае обозначается через $\xi_j$. В задании предполагается распределение Парето вида
$$
\xi_j = (1 - U_j)^{-1/k}, \qquad U_j \sim U(0,1),
$$
где параметр $k > 0$ определяет "тяжесть хвоста" распределения. При $k > 1$ математическое ожидание конечное и равно
$$
\mathbb{E}[\xi] = \frac{k}{k-1}.
$$

При $k \le 1$ средний ущерб бесконечен, что приводит к неминуемому разорению при длительном времени наблюдения.


Обозначим через $K(t)$ капитал компании в момент времени $t$. Он моделируется формулой
$$
K(t) = W + ct - \sum_{t_{(j)} \le t} \, \xi_j,
$$
где
- $W$ — начальный капитал,
- $c$ — скорость притока средств (страховые взносы),
- $\xi_j$ — величина $j$-го страхового ущерба,
- суммируются ущербы всех событий, произошедших до момента $t$.


Средний убыток в единицу времени равен
$$
\lambda \, \mathbb{E}[\xi] = 
\lambda \, \frac{k}{k-1}.
$$

Сравнивая его с притоком средств $c$, получаем три режима работы компании:

- Устойчивый режим
  $$
  c > \lambda \frac{k}{k-1},
  $$
  когда средний приток превышает средний отток — капитал компании растёт.

- Неустойчивый режим
  $$
  c < \lambda \frac{k}{k-1},
  $$
  когда убытки в среднем превышают доходы — капитал убывает и рано или поздно наступает разорение.

- Критический режим 
  $$
  c = \lambda \frac{k}{k-1},
  $$
  граница между устойчивостью и разорением.

Если же $k \le 1$, то $\mathbb{E}[\xi] = \infty$, и разорение неизбежно независимо от величины $c$.


Аналогично предыдущим пунктам, события можно представить в виде двумерного поля точек:
$$
(t_{(j)}, \, \xi_j), \qquad j = 1, \ldots, \eta,
$$
где первая координата — момент страхового случая, вторая — величина ущерба. Это двумерное поле служит основой для построения процесса $K(t)$.


```python
def simulate_insurance(W, c, lam, k, T, n, r):
    if r is not None:
        np.random.seed(r)
    eta = np.random.poisson(lam * T)
    if eta == 0:
        t_events = np.array([])
        xi = np.array([])
        t_grid = np.linspace(0, T, n)
        K_t = W + c * t_grid
        ruin_time = None
        if np.any(K_t < 0):
            ruin_index = np.where(K_t < 0)[0][0]
            ruin_time = t_grid[ruin_index]
        return t_events, xi, t_grid, K_t, ruin_time

    t_events = np.random.uniform(0, T, size=eta)
    t_events = np.sort(t_events)

    U = np.random.uniform(0, 1, size=eta)
    xi = (1 - U) ** (-1 / k)
    t_grid = np.linspace(0, T, n)
    xi_cumsum = np.cumsum(xi)
    idx = np.searchsorted(t_events, t_grid, side="right")
    claims_sum = np.zeros_like(t_grid)
    mask_positive = idx > 0
    claims_sum[mask_positive] = xi_cumsum[idx[mask_positive] - 1]

    K_t = W + c * t_grid - claims_sum
    ruin_time = None
    if np.any(K_t < 0):
        ruin_index = np.where(K_t < 0)[0][0]
        ruin_time = t_grid[ruin_index]

    return t_events, xi, t_grid, K_t, ruin_time


T = 100
lam = 0.5
k = 2

W_good = 50
c_good = 2

W_bad = 20
c_bad = 0.5

t_events_good, xi_good, t_grid_good, K_good, ruin_good = simulate_insurance(
    W=W_good, c=c_good, lam=lam, k=k, T=T, n=2000, r=1
)

t_events_bad, xi_bad, t_grid_bad, K_bad, ruin_bad = simulate_insurance(
    W=W_bad, c=c_bad, lam=lam, k=k, T=T, n=2000, r=2
)

plt.figure(figsize=(12, 8))
plt.subplot(2, 1, 1)
plt.plot(t_grid_good, K_good, label="K(t)")
plt.axhline(0, color="black", linewidth=1)
plt.xlabel("t")
plt.ylabel("K(t)")
plt.title("Капитал страховой компании (устойчивый режим)")
plt.grid(True, linestyle="--", alpha=0.5)

if ruin_good is not None:
    plt.axvline(ruin_good, linestyle="--", label=f"Разорение t ≈ {ruin_good:.2f}")
plt.legend()


plt.subplot(2, 1, 2)
plt.plot(t_grid_bad, K_bad, label="K(t)", color="red")
plt.axhline(0, color="black", linewidth=1)
plt.xlabel("t")
plt.ylabel("K(t)")
plt.title("Капитал страховой компании (разорение)")
plt.grid(True, linestyle="--", alpha=0.5)

if ruin_bad is not None:
    plt.axvline(ruin_bad, linestyle="--", label=f"Разорение t = {ruin_bad:.2f}")
plt.legend()

plt.tight_layout()
plt.show()
```
